# 02 Training Curves

Inspect loss and throughput from a reproducible run in `experiments/runs/`.


```python
import json
from pathlib import Path
root = Path.cwd()
if not (root / 'experiments').exists(): root = root.parent
run = root / 'experiments/runs/wikipedia_30m_smoke_v3'
metrics = run / 'metrics.jsonl'
if metrics.exists():
    rows = [json.loads(line) for line in metrics.read_text(encoding='utf-8').splitlines()]
    for row in rows[-10:]:
        print({key: row.get(key) for key in ('step', 'train_loss', 'validation_loss', 'tokens_per_second')})
else:
    print('Set run to a directory containing metrics.jsonl.')
```
